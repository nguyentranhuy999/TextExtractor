# Khởi động RotoWire benchmark
Chạy trong Colab/Kaggle với Python 3.11 sau khi đưa toàn bộ thư mục dự án vào phiên làm việc. Notebook chỉ gọi CLI chung; dữ liệu và logic thí nghiệm nằm trong rotowire_bench. Đổi đường dẫn PROJECT bên dưới. Môi trường khác phải chạy pilot và dùng run_id mới.

```python
from pathlib import Path
import os, sys, subprocess
PROJECT = Path('/kaggle/working/Text2SQL')  # Colab: /content/Text2SQL
os.chdir(PROJECT)
print(sys.version)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-r', 'requirements-bench.txt'], check=True)
```

```python
def bench(*args):
    subprocess.run([sys.executable, '-m', 'rotowire_bench', *args], check=True)
bench('prepare')
bench('validate')
bench('run', '--dry-run')
```

Máy chạy cần Codex CLI đăng nhập bằng ChatGPT và quyền GPT 5.5. Không dùng API key. Kaggle/Colab không thừa hưởng phiên đăng nhập máy cá nhân; cài CLI, đăng nhập theo tài liệu và chạy preflight/pilot, hoặc export-tasks và nhập câu trả lời UI. Không chép bí mật đăng nhập vào notebook.

```python
# Sau khi Codex CLI đã đăng nhập ChatGPT trên máy này:
bench('pilot', '--limit', '3', '--run-id', 'notebook-pilot-001')
bench('report', '--run-dir', 'outputs/notebook-pilot-001')
```
